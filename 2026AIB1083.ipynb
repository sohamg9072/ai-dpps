### ▶ **Loops**

Suppose you have some task that is **to be done repeatedly**

How will you do that ? Loops

In python, we majorly use two types of loops - **While Loop & For Loop**

While loop is generally used when you do not know the terminating condition.
So you allow the program to check the condition.

For loop is used when you already know when to stop or when there is a known limit.

image.png

```python
#While Loop
start = ""
#You already know that you can add strings from DPP 1
while(start!="aaaaa"):
  start = start + "a"
  print(start)
print("After exiting the while loop, the final Start = ",start)
```

```python
#For Loop
limit = 5
start = 0
sum = 0
for i in range(start, limit):
  sum = sum + i
  print(i)
print("After exiting the while loop, the final Start = ",sum)
```

### ▶ **List, Tuple, Dictionary**

**List**

A List stores elements of same datatype in a sequential order.
The elements can be modified and repeated.

The same indexing strategy is followed as we just learnt in the strings section above.

Some of the list operations are
*   List Creation
*   Update
*   Append
*   Insert
*   Remove
*   Pop

```python
numlist = [1,2,3,5,4,2] #use square brackets
print(numlist)
print(numlist[3])
print(numlist[2:5]) #remember 3,5,4 print hoga
print(numlist[-2])
#'''
stringlist = ['daffy','snoopy','sylvester','tweety']
print(stringlist)
print(stringlist[-1])
#'''
#modification change the list element
stringlist[2] = 'bugsbunny'
print(stringlist)
#'''
#adding elements to the list
#the element gets added in the end
stringlist.append('jaggu')
print(stringlist)
#'''
#Insert at a particular location
subjects = ['maths','science','sst','gk','language','art']
print(subjects)
subjects.insert(2,'sports') #after this 2nd index ka element is sports that is before sst
subjects.insert(5,'music') #before art
print(subjects)
#'''
#Remove a particular item
subjects.remove('sst')
print(subjects)
#'''
#Index wise removal via pop
i = subjects.pop(3)
print(subjects, i)
#'''
#or if no index is given
# it removes the last element
sub = subjects.pop()
print(subjects, sub)
#'''
```

**Tuple**

Tuples are used to store multi-dimensional variables.

A related example for you guys, can be a point **P** in the coordinate geometry stored as (x,y,z).

Similar to the list, it has an order of elements. But, it can not be changed after initialization.

It follows the same indexing procedure as in a list.

Some of the Tuple operations could be :
*   Access
*   Unpack
*   Join
*   Update



```python
Fruittuple = ('guava','grape','pear','mango')
print(Fruittuple)

print(Fruittuple[2])
```

```python
#unpack the existing tuple into other elements
a,b,c,d = Fruittuple
# it is just reverse of initialization
print(a)
print(b)
print(c)
print(d)
```

```python
#join 2 tuples
vegetables = ('brocolli','cauliflower','bittergourd')
eatthis = Fruittuple + vegetables
print(eatthis)

#multiply tuple
print(vegetables*3)
```

```python
#There is no direct way to update tupples.
# we need to first convert it into list and then update it there and then repack it as a tuple

Fruitlist = list(Fruittuple)
print(Fruitlist)

Fruitlist[3] = 'litchi'
print(Fruitlist)
Fruittuple = tuple(Fruitlist) #overwritting value of variable fruittuple
print(Fruittuple)
```

**Dictionary**

Dictionary is just like a real dictionary.

Word : Meaning (Real Dictionary)

Key : Value (Python Dictionary)

Elements of dictionary are ordered and modifiable. But they are unique, there is no repetition.

In a real dictionary also, the words are indexed and one word have a unique meaning.

One key has one particular value only.

Earlier in lists and tuples, the datatype of elements were same. But here the datatype of the elements can be different.

image.png

Below are some of the important operations that can be done with dictionaries


*   Access
*   Checking Existence
*   Update
*   Add
*   Pop

```python
#use curly braces for dictionary
dictionary = {'Subject':"AI", 'Students':360, 'Exams':True, 'LabHours':2}
print(dictionary)
```

```python
#Access
# Careful - Check the [] square brackets
print(dictionary['Students'])
print(dictionary['Exams'])

#print all elements
print(dictionary.items())
#print all values
print(dictionary.values())
#print all keys
print(dictionary.keys())
```

```python
#Checking if the key exists or not
if 'Exams' in dictionary:
  print("Yes")
else:
  print("No")
```

```python
#Values can be changed
dictionary = {'Subject':"AI", 'Students':360, 'Exams':True, 'LabHours':2}
dictionary["Students"] = 180
print(dictionary)

#New entries can be added
dictionary["Teachers"] = 5
print(dictionary)

#entries can be removed by pop
dictionary.pop("Exams")
print(dictionary)
```

**Q16: Ask the user to enter two points in 3D space (x1,y1,z1) and (x2,y2,z2) in the form of tuples. Find the distance between them using the formula you already know from your school.**

```python
point1 = tuple(map(float, input("Enter first point (x1,y1,z1): ").split(',')))
point2 = tuple(map(float, input("Enter second point (x2,y2,z2): ").split(',')))
dx = point2[0] - point1[0]
dy = point2[1] - point1[1]
dz = point2[2] - point1[2]
distance = (dx**2 + dy**2 + dz**2) ** 0.5
print("Distance between points:", distance)
```
