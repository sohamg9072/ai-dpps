# Task 6: Students Who Performed Well
Find all students whose marks are greater than 75.
Don't manually check each student.
* Use Boolean filtering

````python
a = marks.mean(axis=1)          # average per student
good_students = np.where(a > 75)[0]   # indices of students with avg > 75 # obswerve [0] after np.where
print("Students with average > 75:", good_students)
print("Their averages:", a[good_students])
````

# SELF

### Why `np.where(a > 75)[0]` is used

- `np.where(condition)` returns a **tuple of arrays**, one for each dimension of the input.
- Since `a` is a **1D array** (student averages), the result looks like:
  ```python

# Task 14 — Final Challenge 
Create a performance report using NumPy.

* Student 1 → Excellent / Good / Needs Improvement

* Average	Performance
* ≥ 85	Excellent
* ≥ 70 and < 85	Good
