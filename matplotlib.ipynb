# 10. Histogram
* How are the marks distributed

```python
marks=[45,67,32,89,56,72,41,42,43,43,47,90,63,55]
plt.hist(marks)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")
plt.show
```

**Histogram tells us how many observations fall into different ranges**
### For example:
* 0–20       → 0 students
* 20–40      → 1 student
* 40–60      → 4 students
* 60–80      → 3 students
* 80–100     → 2 students

**Y-Axis= frequency=number of data points/students in that range**

# 11. Control the Number of Bins

```python
plt.hist(marks, bins=20)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")
plt.show()
```
