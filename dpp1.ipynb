## Type conversion

Type conversion (also called type casting) is the process of changing a variable from one data type into another.

In Python, this is essential because certain operations require specific data types. For example, you cannot add a string "5" to an integer 10 without converting one of them first.

Python handles type conversion in two ways:

Implicit Conversion: Python automatically changes the data type behind the scenes without your intervention.

Explicit Conversion: You manually force Python to change the data type using built-in functions like int(), float(), str(), list(), etc

```python
age = 25  # This is an integer (number)

# Convert the number to text
age_text = str(age)

print("Your age is " + age_text)
print(type(age_text))
```

```python
text_number = "10"  # This is a string (text)


real_number = int(text_number)   # type conversion

print(real_number + 5)
print(type(real_number))
```

```python
price_text = "9.99"  # This is a string (text)

# Convert the text into a decimal number
price_number = float(price_text)

print(price_number * 2)
print(type(price_number))
```

```python
pi = 3.99  # This is a float (decimal)

# Convert to integer (chops off the decimal part)
whole_number = int(pi)

print(whole_number)
print(type(whole_number))
```

## 1. Arithmetic Operators

Arithmetic operators are symbols used in Python to perform mathematical calculations like addition (\(+\)), subtraction (\(-\)), multiplication (\(*\)), and division (\(/\)). Python features seven core arithmetic operators that evaluate expressions using standard mathematical order of operations (PEMDAS).

```python
a = 10
b = 5
print(a+b) # Addition
```

```python
a = 15
b = 4

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
```

```python
a = 10
b = 3

print(a + b)   # Addition       → 13
print(a - b)   # Subtraction    → 7
print(a * b)   # Multiplication → 30
print(a / b)   # Division       → 3.3333333333333335  (always gives float)
print(a // b)  # Floor Division → 3   (drops the decimal part)
print(a % b)   # Modulus        → 1   (remainder after division)
print(a ** b)  # Exponent       → 1000  (10 raised to the power 3)
```

## 2. Comparison (Relational) Operators

These always return True or False.

```python
a = 10
b = 3

print(a == b)  # Equal to            → False
print(a != b)  # Not equal to        → True
print(a > b)   # Greater than        → True
print(a < b)   # Less than           → False
print(a >= b)  # Greater or equal to → True
print(a <= b)  # Less or equal to    → False
```

## 3. Logical Operators

Used to combine conditions (works with True/False values).

```python
x = 10
y = 5

print(x > 5 and y > 5)   # AND  → False (both must be True)
print(x > 5 or y > 5)    # OR   → True  (at least one must be True)
print(not(x > 5))        # NOT  → False (reverses the result)
```

## 4. Assignment Operators

Used to assign and update variable values.

```python
a = 10
print(a)
a += 5
print(a)  # same as a = a + 5   → 15
a -= 3   # same as a = a - 3   → 12
print(a)
a *= 2   # same as a = a * 2   → 24
print(a)
a /= 4   # same as a = a / 4   → 6.0
print(a)
a //= 2  # same as a = a // 2  → 3.0
print(a)
a **= 2  # same as a = a ** 2  → 9.0
print(a)
a %= 4   # same as a = a % 4   → 1.0
print(a)
```

## 5. Bitwise Operators

Work directly on the binary representation of numbers.

```python
a = 5   # binary: 0101
b = 3   # binary: 0011

print(a & b)   # AND         → 1
print(a | b)   # OR          → 7
print(a ^ b)   # XOR         → 6
print(~a)      # NOT         → -6
print(a << 1)  # Left shift  → 10
print(a >> 1)  # Right shift → 2
```
