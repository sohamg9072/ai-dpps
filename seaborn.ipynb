```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    "Name":["Aman","Priya", "Rahul", "Neha", "Arjun", "Simran", "Karan", "Meena"],
    "Python":[78,65, 88,55,92,74, 61,85],
    "Maths":[85,70,92,60,89,80,67,90],
    "AI":[90,72,95,58,94,82,70,88]

}
df=pd.DataFrame(data)
```

```python
plt.bar(df["Name"],df["Python"])
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.title("Python course performance")
plt.show()
```

```python
sns.barplot(data=df,x="Name",y="Python")
plt.show()
```

```python
df["Gender"] = [
    "Male", "Female", "Male", "Female",
    "Male", "Female", "Male", "Female"
]
```

```python
sns.scatterplot(
    data=df,
    x="Python",
    y="AI",
    hue="Gender"
)

plt.show()
```
