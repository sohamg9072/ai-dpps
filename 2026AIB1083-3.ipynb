## **Basic Error Handling**

> **Basic error handling is a way of dealing with errors in a program so that the program does not stop suddenly and can respond properly when something goes wrong.**

###  **Error handling = Telling Python what to do when something goes wrong.**

For example, if a program asks the user to enter a number but the user enters `"hello"`, Python may produce an error. **Error handling** allows us to handle such situations properly instead of letting the program stop unexpectedly.


# **Example:- Write a python program for Basic Error Handling.**

Suppose we ask the user to enter a number. If the user enters text instead of a number, the program can produce an error.

We can handle this using **`try` and `except`**:

* **`try`** → contains the code that we want Python to try running.
* **`except`** → tells Python what to do if an error occurs.
* **`ValueError`** → occurs when the user enters something that cannot be converted into an integer.



````python
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)

except ValueError:
    print("Please enter a valid number.")
````

# **Q11. Write a Python program using functions to create a menu-driven calculator for different calculations.**

The program should allow the user to choose one of the following options:

1. **Circle** – Calculate area and circumference.
2. **Rectangle** – Calculate area and perimeter.
3. **Triangle** – Calculate area.
4. **Simple Interest** – Calculate simple interest.
5. **Speed** – Calculate speed.
6. **Average** – Calculate the average of three numbers.
7. **Grade** – Calculate the grade from marks.

### **Conditions**

- Create a **separate function for each calculation**.
- Each function must take the required values as **parameters**.
- Each function must **return** the result.
- Display a **menu** and take the user's choice.
- Take the required input values based on the selected option.
- Use `math.pi` for circle calculations.
- Display the calculated result clearly.
- For the **Grade** option, return the appropriate grade based on the marks.

### **Example**

```text
===== Calculation Menu =====

1. Circle
2. Rectangle
3. Triangle
4. Simple Interest
5. Speed
6. Average
7. Grade

Enter your choice: 1
Enter radius: 7

Area = 153.94
Circumference = 43.98

````python
def circle_area(radius):
    return 3.14159 * radius * radius

def rectangle_area(length, width):
    return length * width

def triangle_area(base, height):
    return 0.5 * base * height

def simple_interest(principal, rate, time):
    return (principal * rate * time) / 100

def speed(distance, time):
    return distance / time

def average(numbers):
    return sum(numbers) / len(numbers)

def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 40:
        return "D"
    else:
        return "F"


# Menu display
def display_menu():
    print("\n--- Calculator Menu ---")
    print("1. Circle Area")
    print("2. Rectangle Area")
    print("3. Triangle Area")
    print("4. Simple Interest")
    print("5. Speed")
    print("6. Average")
    print("7. Grade")
    print("8. Exit")



results = []  # to store results 

while True:
    display_menu()
    choice = int(input("Enter your choice: "))

    if choice == 1:
        r = float(input("Enter radius: "))
        res = circle_area(r)
    elif choice == 2:
        l = float(input("Enter length: "))
        w = float(input("Enter width: "))
        res = rectangle_area(l, w)
    elif choice == 3:
        b = float(input("Enter base: "))
        h = float(input("Enter height: "))
        res = triangle_area(b, h)
    elif choice == 4:
        p = float(input("Enter principal: "))
        r = float(input("Enter rate: "))
        t = float(input("Enter time: "))
        res = simple_interest(p, r, t)
    elif choice == 5:
        d = float(input("Enter distance: "))
        t = float(input("Enter time: "))
        res = speed(d, t)
    elif choice == 6:
        nums = list(map(float, input("Enter numbers separated by space: ").split()))
        res = average(nums)
    elif choice == 7:
        m = float(input("Enter marks: "))
        res = grade(m)
    elif choice == 8:
        print("Exiting program...")
        break
    else:
        print("Invalid choice!")
        continue

    print("Result:", res)


    if True:
        results.append(res)

# Display stored results
print("\nStored Results (first three calculations):", results)
````

- Ask the user for the required password length.

- If the user enters an invalid length:

  - Allow the user to enter the length again.

- After generating the password, create a separate validation function that checks **each password condition**.

- The validation function should return:
  - `Valid Password` if all conditions are satisfied.
  - `Invalid Password: <reason>` if any condition is not satisfied.

- The validation function should **not directly print the result**.

- If the generated password does not satisfy the required conditions, generate another password until a valid password is produced.

### **Example 1: Valid Password**

**Input:**

```text
Enter password length: 10

OUTPUT

Generated Password = A7@kp2#Mx9

Status = Valid Password

````python
import random
import string

# ------------------ GENERATORS ------------------
def random_uppercase():
    return random.choice(string.ascii_uppercase)

def random_lowercase():
    return random.choice(string.ascii_lowercase)

def random_digit():
    return random.choice(string.digits)

def random_special():
    specials = "!@#$%^&*()-_=+[]{};:,.<>?/|"
    return random.choice(specials)

# ------------------ PASSWORD BUILDER ------------------
def generate_password(length):
    # Ensure password has at least one of each type
    password = [
        random_uppercase(),
        random_lowercase(),
        random_digit(),
        random_special()
    ]

    # Fill the rest randomly from all groups
    all_chars = string.ascii_letters + string.digits + "!@#$%^&*()-_=+[]{};:,.<>?/|"
    while len(password) < length:
        password.append(random.choice(all_chars))

    # Shuffle to avoid predictable order
    random.shuffle(password)
    return "".join(password)

# ------------------ VALIDATOR ------------------
def validate_password(password):
    if len(password) < 8:
        return "Invalid Password: Length must be at least 8 characters."

    if not any(c.isupper() for c in password):
        return "Invalid Password: Must contain at least one uppercase letter."
    if not any(c.islower() for c in password):
        return "Invalid Password: Must contain at least one lowercase letter."
    if not any(c.isdigit() for c in password):
        return "Invalid Password: Must contain at least one digit."
    if not any(c in "!@#$%^&*()-_=+[]{};:,.<>?/|" for c in password):
        return "Invalid Password: Must contain at least one special character."

    return "Valid Password"

# ------------------ MAIN PROGRAM ------------------
def main():
    while True:
        try:
            length = int(input("Enter password length: "))
            if length < 8:
                print("Error: Password length must be at least 8.")
                continue
            break
        except ValueError:
            print("Error: Please enter a valid number.")

    # Keep generating until valid
    while True:
        password = generate_password(length)
        status = validate_password(password)
        if status == "Valid Password":
            print("\nGenerated Password =", password)
            print("Status =", status)
            break

# Run program
if __name__ == "__main__":
    main()
````

````python
from datetime import datetime

# ------------------ TIME CONVERSION ------------------
def convert_time(time_str):
    try:
        # Convert "HH:MM" string into datetime object (only time part matters)
        return datetime.strptime(time_str, "%H:%M")
    except ValueError:
        return None

# ------------------ CONFLICT CHECK ------------------
def check_conflict(start1, end1, start2, end2):
    # Meetings overlap if one starts before the other ends AND ends after the other starts
    return start1 < end2 and start2 < end1

# ------------------ RESULT GENERATOR ------------------
def generate_result(conflict):
    if conflict:
        return "Conflict Detected\nThe meetings overlap."
    else:
        return "No Conflict\nThe meetings do not overlap."

# ------------------ MAIN PROGRAM ------------------
def main():
    while True:
        try:
            m1_start = input("Enter Meeting 1 start time (HH:MM): ")
            m1_end = input("Enter Meeting 1 end time (HH:MM): ")
            m2_start = input("Enter Meeting 2 start time (HH:MM): ")
            m2_end = input("Enter Meeting 2 end time (HH:MM): ")

            start1 = convert_time(m1_start)
            end1 = convert_time(m1_end)
            start2 = convert_time(m2_start)
            end2 = convert_time(m2_end)

            if None in (start1, end1, start2, end2):
                raise ValueError("Invalid time format. Please use HH:MM.")

            if start1 >= end1 or start2 >= end2:
                raise ValueError("End time must be after start time.")

            conflict = check_conflict(start1, end1, start2, end2)
            print("\n" + generate_result(conflict))
            break

        except ValueError as e:
            print("Error:", e)
            print("Please try again.\n")

main()
````
